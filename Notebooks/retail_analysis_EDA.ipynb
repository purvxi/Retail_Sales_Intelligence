```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
```

```python
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['Price']
```

```python
df_clean[['InvoiceDate', 'Quantity', 'Price', 'Revenue']].head()
```

```python
country_revenue = (
    df_clean.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

country_revenue.head(10)
```

```python
country_analysis = df_clean.groupby('Country').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Transactions=('Invoice', 'nunique'),
    Unique_Customers=('Customer ID', 'nunique')
)
```

```python
country_analysis['Avg_Transaction_Value'] = (
    country_analysis['Total_Revenue'] /
    country_analysis['Total_Transactions']
)
```

```python
country_analysis.sort_values(
    by='Total_Revenue', ascending=False
).head(10)
```

**Observations**
The UK contributes the highest total revenue and transaction volume.

The Netherlands has a considerably higher average transaction value despite lower total revenue.

Transaction volume and identifiable customer count vary across countries.

Customer-level comparisons are limited by missing Customer IDs.

```python
product_analysis = df_clean.groupby('Description').agg(
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('Revenue', 'sum'),
    Total_Transactions=('Invoice', 'nunique')
)
```

```python
product_analysis.sort_values(
    by='Total_Revenue', ascending=False
).head(10)
```

```python
product_analysis.sort_values(
    by='Total_Quantity', ascending=False
).head(10)
```

```python
# Bottom 10 products by revenue
product_analysis.sort_values(
    by='Total_Revenue', ascending=True
).head(10)

# Bottom 10 products by quantity
product_analysis.sort_values(
    by='Total_Quantity', ascending=True
).head(10)
```

```python
customer_analysis = df_clean.dropna(subset=['Customer ID']).groupby('Customer ID').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Orders=('Invoice', 'nunique'),
    Total_Quantity=('Quantity', 'sum')
)
```

```python
customer_analysis.sort_values(
    by='Total_Revenue', ascending=False
).head(10)
```

**Customer 18102** generates the highest revenue, despite not having the highest order count.

**Customer 14646** purchases the highest quantity among these customers.

**Customer 14911** has the highest order frequency (510 orders), but ranks fourth in revenue.

```python
total_customer_revenue = customer_analysis['Total_Revenue'].sum()
top_10_revenue = customer_analysis.nlargest(
    10, 'Total_Revenue'
)['Total_Revenue'].sum()

print("Total identified customer revenue:", total_customer_revenue)
print("Top 10 customer revenue:", top_10_revenue)
print("Top 10 contribution (%):",
      (top_10_revenue / total_customer_revenue) * 100)
```

Conclusion :The retailer has a relatively broad revenue base across its identified customers, rather than relying entirely on a handful of customers.

```python
df_clean['Year'] = df_clean['InvoiceDate'].dt.year
df_clean['Month'] = df_clean['InvoiceDate'].dt.month
df_clean['YearMonth'] = df_clean['InvoiceDate'].dt.to_period('M')

monthly_analysis = df_clean.groupby('YearMonth').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Transactions=('Invoice', 'nunique')
)

monthly_analysis
```

The retailer appears to experience stronger purchasing activity during the September–November period, potentially reflecting seasonal and holiday-related demand.

```python
product_sales = df_clean[
    (df_clean['Quantity'] > 0) &
    (df_clean['Price'] > 0)
].copy()

product_sales = product_sales[
    ~product_sales['Description'].str.upper().isin(
        ['POSTAGE', 'DOTCOM POSTAGE']
    )
]
```

```python
top_products = (
    product_sales.groupby('Description')['Quantity']
    .sum().sort_values(ascending=False).head(5)
)
top_products
```

```python
top_product_trends = product_sales[
    product_sales['Description'].isin(top_products.index)
].groupby(['YearMonth', 'Description'])['Quantity'].sum().unstack(fill_value=0)

top_product_trends
```

```python
white_heart = product_sales[
    product_sales['Description'] ==
    'WHITE HANGING HEART T-LIGHT HOLDER'
]

white_heart_monthly = white_heart.groupby(
    ['Year', 'Month']
)['Quantity'].sum().unstack()

white_heart_monthly
```

```python
white_heart_monthly_avg = (
    white_heart.groupby('Month')['Quantity']
    .mean()
    .sort_values(ascending=False)
)

white_heart_monthly_avg
```

The monthly product demand analysis revealed variations in purchasing patterns across the five highest-volume products. While some products demonstrated lower sales quantities during several months of 2011 compared with 2010, others experienced fluctuating demand and occasional sales peaks. These variations suggest that product demand may be influenced by seasonal purchasing patterns and bulk transactions. Products showing declining demand should therefore be monitored before making inventory reduction or discontinuation decisions. Seasonal inventory planning and further profitability assessment can help retailers make better product-retention decisions.

One limitation: We cannot establish actual profitability or ROI because the dataset does not contain product costs or inventory-holding costs.
